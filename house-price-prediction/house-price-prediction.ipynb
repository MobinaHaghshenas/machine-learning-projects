## 3.Feature Selection & preprocessing


```python
from sklearn.preprocessing import LabelEncoder

# Split the data into features and target variable
X = df.drop('SalePrice', axis=1)
y = df['SalePrice']

# Create a copy of X
X_encoded = X.copy()

# Identify categorical features
categorical_features = X.select_dtypes(include=['object']).columns

# Encode categorical features using Label Encoding
for feature in categorical_features:
    le = LabelEncoder()
    X_encoded[feature] = le.fit_transform(X_encoded[feature])

# Calculate correlation coefficients
corr_coeffs = X_encoded.corrwith(y)


results = pd.DataFrame({"Feature": corr_coeffs.index, "Correlation Coefficient": corr_coeffs.abs()})
results = results.sort_values("Correlation Coefficient", ascending=False).reset_index(drop=True)

# Select features
important_features = results[(results["Correlation Coefficient"] > 0.50) | (results["Correlation Coefficient"] < -0.50)]["Feature"].tolist()


original_features = df.columns.tolist()
important_features_filtered = [feature for feature in important_features if feature in original_features]


print("Selected Features:")
print(important_features_filtered)


df2 = df[important_features_filtered + ['SalePrice']]

```

* I selected  features which have more than 0.50 or less than -0.50 correlation rate.

```python
df2.head()
```

```python
# Removing missing values
df2.dropna(inplace=True)
print("TOTAL MISSING VALUES:",df2.isna().sum().sum())
```

```python
#Remove Outliers from numerical features
df_numeric = df2.select_dtypes(include=[np.number])
Q1 = df_numeric.quantile(0.25)
Q3 = df_numeric.quantile(0.75)
IQR = Q3 - Q1
df2 = df2[~((df2[df_numeric.columns] < (Q1 - 1.5 * IQR)) | (df2[df_numeric.columns] > (Q3 + 1.5 * IQR))).any(axis=1)]
```

## 5.Preliminary Assessment of Linear Models


```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor

```

***Defining evaluation functions***

```python
def rmse_cv(model):
    rmse = np.sqrt(-cross_val_score(model, X, y, scoring="neg_mean_squared_error", cv=5)).mean()
    return rmse


def evaluation(y, predictions):
    mae = mean_absolute_error(y, predictions)
    mse = mean_squared_error(y, predictions)
    rmse = np.sqrt(mean_squared_error(y, predictions))
    r_squared = r2_score(y, predictions)
    return mae, mse, rmse, r_squared
```

```python
Models = pd.DataFrame(columns=["Model","MAE","MSE","RMSE","R2 Score","RMSE (Cross-Validation)"])
```

 <h2 style="font-family: 'Times New Roman'; letter-spacing: 0.05em;">Linear Regression</h2>

```python
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)
predictions_Linear_Regression = lin_reg.predict(X_test)

mae, mse, rmse, r_squared = evaluation(y_test, predictions_Linear_Regression)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r_squared)
print("-"*30)
rmse_cross_val = rmse_cv(lin_reg)
print("RMSE Cross-Validation:", rmse_cross_val)

new_row = {"Model": "LinearRegression","MAE": mae, "MSE": mse, "RMSE": rmse, "R2 Score": r_squared, "RMSE (Cross-Validation)": rmse_cross_val}
Models = pd.concat([Models, pd.DataFrame([new_row])], ignore_index=True)
```

```python
def plot_predictions(models, predictions):


  num_rows = len(models) // 2 + len(models) % 2
  num_cols = 2

  fig, axes = plt.subplots(num_rows, num_cols, figsize=(12, 8))


  for i, ax in enumerate(axes.flat):
      if i < len(models):
          ax.scatter(y_test, predictions[i], color='blue')
          ax.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], 'r--', lw=2)
          ax.set_title(models[i])
      else:
          fig.delaxes(ax)

  plt.tight_layout()
  plt.show()

models = ['Linear_Regression', 'Ridge_Regression', 'Lasso_Regression', 'XGBoost', 'RandomForest', 'GradientBoosting', 'DecisionTree', 'AdaBoost']
predictions = [predictions_Linear_Regression, predictions_Ridge_Regression, predictions_Lasso_Regression, predictions_XGB, predictions_RandomForest, predictions_GradientBoosting, predictions_DecisionTree, predictions_AdaBoost]

plot_predictions(models, predictions)
```

```python
def plot_predictions(models, y_actual, y_predictions):
    num_rows = len(models) // 3 + (len(models) % 3 > 0)
    num_cols = 3

    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 10))
    axes = axes.flatten()

    for i, ax in enumerate(axes):
        if i < len(models):

            ax.scatter(range(len(y_actual)), y_actual, color='blue', label='Actual', alpha=0.6)
            ax.scatter(range(len(y_predictions[i])), y_predictions[i], color='red', label='Predicted', alpha=0.6)

            ax.set_title(models[i])
            ax.legend()
        else:
            ax.axis('off')

    plt.tight_layout()
    plt.show()

models = ['Linear_Regression', 'Ridge_Regression', 'Lasso_Regression', 'XGBoost', 'RandomForest', 'GradientBoosting', 'DecisionTree', 'AdaBoost']
predictions = [predictions_Linear_Regression, predictions_Ridge_Regression, predictions_Lasso_Regression, predictions_XGB, predictions_RandomForest, predictions_GradientBoosting, predictions_DecisionTree, predictions_AdaBoost]


plot_predictions(models, y_test, predictions)
```
